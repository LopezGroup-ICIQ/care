# Adsorbate placement with CARE

This notebooks aims at explaining to the user how adsorbate placement, a mandatory step in any atomistic study in heterogenous catalysis, is performed within CARE. <br/>

The main dependence working behind the scenes of this step is a modified version of DockOnSurf. <br/>

With CARE, intermediates within the reaction networks are automatically placed on the surface. If the framework is used as a pipeline, the built-in functions are integrated in the intermediate evaluation but the functions are also available as independent modules. To show what's behind the code, we showcase how CARE handles the generation of different adsorption configurations.

### Step 1: Define the adsorbate

⏲ 0.2s (Ubuntu), 0.3s (MacOS)

````python
from care import Intermediate
from ase.visualize.plot import plot_atoms
````

````python
adsorbate = Intermediate.from_smiles("CO")
print(adsorbate)
````

````python
plot_atoms(adsorbate.molecule, rotation=('0x, 0y, 0z'), show_unit_cell=0)
````

### Step 2: Define the surface of interest

As example we will consider here a simple metal surface. To know how to load more complex surface, please check the notebook ``care_demo.ipynb``

````python
from care import Surface
````

````python
surface = Surface.from_metal_db(metal="Au", hkl="111")
plot_atoms(surface.slab, rotation="0x, 0y, 90z", show_unit_cell=2)
````

### Step 3: Perform adsorbate placement

The function ```place_adsorbate``` recieves as inputs the ```Intermediate``` object of the desired adsorbate, the ```Surface``` object of the desired catalyst and the number of desired configurations to generate ```num_configs``` as an integer value. If all possible configurations are desired to be generated, the user has to define: ```num_configs=-1```

The output of the function is a list of the different configurations as ```ase.Atoms``` objects: ```list[ase.Atoms]```.

Now, let's place the adsorbate on the surface! The function will automatically identify the potential active sites of the surface and anchoring points of the molecule and will automatically adapt the surface to avoid potential overlap of the adsorbate or lateral interactions.


````python
from care.adsorption import place_adsorbate
````

⏲ 2.8s (Ubuntu), 5.7s (MacOS)

````python
adsorption_configurations = place_adsorbate(intermediate=adsorbate,
                                            surface=surface,
                                            num_configs=10,
)
````

````python
print(f'Number of generated adsorption configurations: {len(adsorption_configurations)}')
print(adsorption_configurations[0])  # Print the first configuration
print(type(adsorption_configurations[0]))  # Check the type of the first configuration
````

````python
plot_atoms(adsorption_configurations[0], rotation='0x, 70y, 90z', show_unit_cell=0)
````
